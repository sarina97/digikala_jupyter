```python
import mysql.connector
import pandas as pd
import time
from IPython.display import clear_output

class DataStream:
    def __init__(self, host, user, password, database,
                 initial_load=1000, batch_size=200, delay=0.5):

        self.initial_load = initial_load
        self.batch_size = batch_size
        self.delay = delay

        # اتصال به دیتابیس
        self.conn = mysql.connector.connect(
            host=host,
            user=user,
            password=password,
            database=database
        )
        self.cursor = self.conn.cursor(dictionary=True)

        # دیتا فریم اصلی
        self.df = pd.DataFrame()
        self.offset = 0

    # گرفتن batch مرتب شده
    def get_batch(self, offset, limit):
        query = f"""
            SELECT *
            FROM products
            ORDER BY id
            LIMIT {limit} OFFSET {offset};
        """

        self.cursor.execute(query)
        data = self.cursor.fetchall()
        return pd.DataFrame(data)

    # استارت stream
    def start_stream(self, live_display=False):
        print("🔄 Loading initial data...")

        # اولین لود
        self.df = self.get_batch(self.offset, self.initial_load)
        self.offset += self.initial_load

        print(f"✔ Loaded {len(self.df)} initial records.")

        # استریم کردن batchها
        while True:
            new_batch = self.get_batch(self.offset, self.batch_size)

            if new_batch.empty:
                print("✔ No more data in SQL — Stream completed.")
                break

            # الحاق به df
            self.df = pd.concat([self.df, new_batch], ignore_index=True)
            self.offset += self.batch_size

            if live_display:
                clear_output(wait=True)
                print("📌 Last update:", pd.Timestamp.now())
                print("📊 Total rows loaded:", len(self.df))
                print("Average discount:", self.df["discount_percent"].astype(float).mean())
                display(self.df.tail(8))

            time.sleep(self.delay)

    def close(self):
        self.cursor.close()
```

```python
from persiantools.jdatetime import JalaliDate
import pandas as pd

df['Date_sell'] = pd.to_datetime(df['Date_sell'], errors='coerce')
def to_jalali(date_str):
    try:
        d = pd.to_datetime(date_str, errors="coerce")
        if pd.isna(d):
            return None
        j = JalaliDate.to_jalali(d.year, d.month, d.day)
        return f"{j.year}/{j.month}/{j.day}"
    except:
        return None

df["persian_date_sell"] = df["Date_sell"].apply(to_jalali)

print(df[["Date_sell", "persian_date_sell"]].head(10))
df.to_csv("digikala_products_117K.csv", index=False)
```

```python
from persiantools import jdatetime
import jdatetime
# df["Date-sell"].dt.year.unique()

def extract_persian_year(x):
    try:
        if '/' in str(x):
            return int(str(x).split('/')[0])
        else:
            gdate = pd.to_datetime(x)
            return jdatetime.date.fromgregorian(date=gdate.date()).year
    except:
        return None

def extract_persian_month(x):
    try:
        if '/' in str(x):
            return int(str(x).split('/')[1])
        else:
            gdate = pd.to_datetime(x)
            return jdatetime.date.fromgregorian(date=gdate.date()).month
    except:
        return None

df['persian_year'] = df['persian_date_sell'].apply(extract_persian_year)
df['persian_month'] = df['persian_date_sell'].apply(extract_persian_month)


df['persian_year'].unique()
print(sorted(df["persian_month"].unique()))
df.to_csv("digikala_products_117K.csv", index=False)
```

```python
# برای اینکه ببینیم ارزانتر یا گرانتر بودن بین دسته ها اتفاقی یا معنا دار
counts = df['item_category2'].value_counts()

df.groupby('item_category2' , observed=True)['selling_price'].count()
df.groupby('item_category2',observed=True)['selling_price'].var()

# print(df['item_category2'].value_counts())
cats = df['item_category2'].value_counts().nlargest(6).index.tolist()
sample = df[df['item_category2'].isin(cats)]

valid_groups = []
valid_names = []
for name, group in sample.groupby('item_category2' , observed=True):
    prices = group['selling_price'].dropna().values
    if len(prices) >= 3 and prices.var() > 0:
        valid_groups.append(prices)
        valid_names.append(name)

if len(valid_groups) >= 2:
    fstat, pval = stats.f_oneway(*valid_groups)
    print("Valid categories:", valid_names)
    print("ANOVA F-stat:", fstat, "p-value:", pval)
else:
    print("گروه‌های کافی برای ANOVA موجود نیستند.")
```

```python
import pandas as pd

# تعریف فصل‌ها از ماه
def get_season_from_month(month):
    if month in [1,2,3]:
        return 'Winter'
    elif month in [4,5,6]:
        return 'Spring'
    elif month in [7,8,9]:
        return 'Summer'
    else:
        return 'Autumn'

# اضافه کردن فصل به df_product
df_product['season'] = df_product['persian_month'].apply(get_season_from_month)

# اعمال fas برای فارسی کردن
df_product['brand'] = df_product['brand'].apply(fas)
df_product['item_category2'] = df_product['item_category2'].apply(fas)

# محاسبه مجموع فروش برای هر برند و آیتم
df_item_total = (
    df_product.groupby(['brand','item_category2'])
    .size()
    .reset_index(name='total_count')
)

# انتخاب ۵ آیتم پرفروش برای هر برند
top_items_per_brand = (
    df_item_total.sort_values(['brand','total_count'], ascending=[True,False])
    .groupby('brand')
    .head(5)
)

# انتخاب ۵ برند پرفروش
top_brands = (
    df_product.groupby('brand')
    .size()
    .reset_index(name='brand_total')
    .sort_values('brand_total', ascending=False)
    .head(5)
)['brand'].tolist()

# تولید Insight
for brand in top_brands:
    print(f"\n🔹 برند: {fas(brand)}")

    top_items = top_items_per_brand[top_items_per_brand['brand']==brand]['item_category2'].tolist()

    for season in ['Spring','Summer','Autumn','Winter']:
        df_season = df_product[
            (df_product['brand']==brand) &
            (df_product['season']==season) &
            (df_product['item_category2'].isin(top_items))
        ]

        if df_season.empty:
            print(f"\nفصل {season}: هیچ فروشی برای آیتم‌های برتر ثبت نشده است.")
            continue

        # تعداد فروش هر آیتم در آن فصل
        df_counts = df_season.groupby('item_category2').size().reset_index(name='count')

        # مجموع فروش کل فصل
        season_total = df_counts['count'].sum()

        # درصد فروش
        df_counts['perc'] = df_counts['count'] / season_total * 100

        print(f"\nفصل {season}:")
        for _, row in df_counts.iterrows():
            print(f"  {fas(row['item_category2'])}: {row['perc']:.1f}% فروش")
```

```python
def get_season_from_month(month):
    if month in [1,2,3]:
        return 'Winter'
    elif month in [4,5,6]:
        return 'Spring'
    elif month in [7,8,9]:
        return 'Summer'
    else:
        return 'Autumn'

def clean_cat(x):
    if isinstance(x, str) and x.startswith("["):
        return x.split(",")[-1].replace("]", "").strip()
    return x


df_product['season'] = df_product['persian_month'].apply(get_season_from_month)

df_product['category_clean'] = df_product['category'].apply(clean_cat)

top3_itemcats = (
    df_product['item_category2']
    .value_counts()
    .head(3)
    .index
    .tolist()
)

for item_cat in top3_itemcats:

    print(f"\n📌 تحلیل فصلی برای: {item_cat}")

    df_item = df_product[df_product['item_category2'] == item_cat]

    for season in ['Spring', 'Summer', 'Autumn', 'Winter']:

        df_season = df_item[df_item['season'] == season]

        if df_season.empty:
            print(f" در فصل {season} هیچ داده‌ای برای {fas(item_cat)} وجود ندارد.")
            continue

        df_counts = (
            df_season.groupby('category_clean')
            .size()
            .reset_index(name='count')
            .sort_values('count', ascending=False)
        )

        top3 = df_counts.head(3).copy()
        total = top3['count'].sum()
        top3.loc[:, 'percent'] = (top3['count'] / total * 100).round(2)

        plt.figure(figsize=(10,4))

        ax = sns.barplot(
            y=[fas(x) for x in top3['category_clean']],
            x=top3['percent'],
            orient='h'
        )

        for i, p in enumerate(top3['percent']):
            ax.text(
                p + 0.3,
                i,
                f"{p}%",
                va='center',
                fontproperties=vazir_font
            )

        plt.title(f"{fas(item_cat)} - {season}", fontproperties=vazir_font)
        plt.xlabel(fas("درصد فروش فصل"), fontproperties=vazir_font)
        plt.ylabel(fas("دسته‌ها"), fontproperties=vazir_font)

        plt.tight_layout()
        plt.show()
```
